```python
from sklearn.datasets import make_classification
import numpy as np
X,y=make_classification(n_samples=100,n_features=2,n_redundant=0,n_classes=2,n_clusters_per_class=1,random_state=41)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter')
```

```python
def step(z):
  return 0 if z<=0 else 1
```

```python
def Perceptron(X,y):
  X=np.insert(X,0,1,axis=1)
  weights=np.ones(X.shape[1])
  lr=0.1

  for i in range(1000):
    j=np.random.randint(0,100)
    y_pred=step(np.dot(X[j],weights))
    weights=weights+lr*(y[j]-y_pred)*(X[j])

  return weights[0],weights[1:]
```

```python
intercept_,coef_=Perceptron(X,y)
print(intercept_)
print(coef_)
```

```python
m=-(coef_[0]/coef_[1])
b=-(intercept_/coef_[1])
```

```python
x_input=np.linspace(-3,3,1000)
y_input=m*x_input+b

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='pink',linewidth=5)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=25)
plt.ylim(-3,2)
```
