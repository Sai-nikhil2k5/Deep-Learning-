```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[9,9,9],[8,8,4],[7,9,5],[6,10,6],[5,12,7]],columns=['iq','cgpa','lpa'])
```

```python
df
```

```python
def init_parameters(layer_dims):
  L=len(layer_dims)
  parameters={}
  for l in range(1,L):
    parameters['W' + str(l)]=np.ones((layer_dims[l-1],layer_dims[l]))*0.1
    parameters['b' + str(l)]=np.zeros((layer_dims[l],1))

  return parameters
```

```python
def linear_forward(A_prev,W,b):
  z=np.dot(W.T,A_prev)+b
  return z
```

```python
def L_layer_forward(X,parameters):
  A=X
  L=len(parameters)//2

  for l in range(1,L+1):
    A_prev=A
    Wl=parameters['W'+str(l)]
    bl=parameters['b'+str(l)]

    A= linear_forward(A_prev,Wl,bl)

    return A,A_prev
```

```python
X=df[['iq','cgpa']].values[0].reshape(2,1)
y=df[['lpa']].values[0][0]

parameters= init_parameters([2,2,1])

y_hat,A_prev=L_layer_forward(X,parameters)
```

```python
A_prev
```

```python
def update_parameters(X,y,y_hat,A_prev,parameters):
  parameters['W2'][0][0]=parameters['W2'][0][0]+(0.001*(y-y_hat)*A_prev[0][0])
  parameters['W2'][1][0]=parameters['W2'][1][0]+(0.001*(y-y_hat)*A_prev[1][0])
  parameters['b2'][0][0]=parameters['b2'][0][0]+(0.001*(y-y_hat))

  parameters['W1'][0][0]=parameters['W1'][0][0]+(0.001*(y-y_hat)*parameters['W2'][0][0]*X[0][0])
  parameters['W1'][0][1]=parameters['W1'][0][1]+(0.001*(y-y_hat)*parameters['W2'][0][0]*X[1][0])
  parameters['b1'][0][0]=parameters['b1'][0][0]+(0.001*(y-y_hat)*parameters['W2'][0][0])

  parameters['W1'][1][0]=parameters['W1'][1][0]+(0.001*(y-y_hat)*parameters['W2'][1][0]*X[0][0])
  parameters['W1'][1][1]=parameters['W1'][1][1]+(0.001*(y-y_hat)*parameters['W2'][1][0]*X[1][0])
  parameters['b1'][1][0]=parameters['b1'][1][0]+(0.001*(y-y_hat)*parameters['W2'][1][0])

```

```python
y_hat = y_hat[0][0]
update_parameters(X,y,y_hat,A_prev,parameters)
```

```python
parameters=init_parameters([2,2,1])

epochs=10

for i in range(epochs):

  Loss=[]

  for j in range(df.shape[0]):

    X=df[['iq','cgpa']].values[j].reshape(2,1)
    y=df[['lpa']].values[j][0] # Explicitly get the scalar value from the array

    y_hat,A_prev=L_layer_forward(X,parameters)
    y_hat=y_hat[0][0]

    update_parameters(X,y,y_hat,A_prev,parameters)

    Loss.append((y-y_hat)**2)

  print('Epoch',i+1,'loss:',np.array(Loss).mean())
```
